### LINKED LISTS
This is the implementation of linked lists, the most probable material that will come in the exam. This implementation is based on the SP pdf file, which will be the fundemental application.

```python
class Node:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next
```

```python
class LinkedList:
    def __init__(self):
        self.front = None

    def __repr__(self):
        s = '['
        node = self.front
        while node is not None:
            s = s + repr(node.value)
            node = node.next
            if node is not None: s += ','
        s += ']'
        return s
        
    def insert(self, value, where=None):
        if where is None:
            node = Node(value, self.front)
            self.front = node # referencing to the next node
        else:
            node = Node(value, where.next)
            where.next = node

    def remove(self, where=None):
        if where is None:
            if self.front is not None:
                self.front = self.front.next
        else:
            if where.next is not None:
                where.next = where.next.next

    def find(self, value):
        node = self.front
        while node is not None:
            if node.value == value:
                return node
            node = node.next
        return print("Node does not found!")

L = LinkedList()
L.insert(5)
L.insert(10)
L.insert(5)
L.insert(10)
L.insert(5)
L.insert(10)
L
L.remove()
L.find(20)
```
