### LIST OF DICTIONARY EXERCISES - PRIVATE EXERCISE¶
This is my exercise for the list of dictionary usage as well as the linked list implementation.

```python
memory = [
    {"val": "Scientific", "next": 2},    # Index 0
    {"val": "Finish",     "next": None}, # Index 1 (End of list)
    {"val": "Programming", "next": 1}    # Index 2
]

list_of_memory = [dict() for _ in range(0, 10)]
head_index = 0

current = head_index

while current is not None:
    node = memory[current]
    print(node["val"])
    # update the node
    current = node["next"]
```

```python
list_of_elements = ["Hydrogen", "Helium", "Lithium"]
atomic_number=[1, 2, 3]
type_gas=["Non Metal", "Noble Gas", "Alkali Metal"]
elements = [dict() for _ in range(len(list_of_elements))]

for n in range(len(list_of_elements)):
    elements[n]["element"] = list_of_elements[n]
    elements[n]["atomic number"] = atomic_number[n]
    elements[n]["Type gas"] = type_gas[n]
    
print(elements)
print(elements[1]["atomic number"])

elements.append({"element" : "Carbide"})
print(elements)
```

```python
memory = {
    "A1": {"chemical": "Ethanol", "next_id": "C3"},
    "C3": {"chemical": "Acetone", "next_id": "B2"},
    "B2": {"chemical": "Water",   "next_id": None}
}

current_id = "A1"
my_node = memory[current_id]
print(my_node["chemical"])

while current_id is not None:
    # access the current_id inside the memory
    node = memory[current_id]
    # print the information in the id, which is chemical
    print(node["chemical"])
    # update the next id
    current_id = node["next_id"]
```

```python
results_memory_test = {
    "node1": {"value": 10.5, "status": "valid",   "next": "node2"},
    "node2": {"value": -999, "status": "invalid", "next": "node3"},
    "node3": {"value": 15.2, "status": "valid",   "next": None}
}

current_node_test = "node1"
def my_generator_function(current_node, results_memory):
    while current_node is not None:
        # access the current node inside the memory result
        node = results_memory[current_node]
        # implement yield function
        if node["status"] == "valid":
            yield node["value"]
        current_node = node["next"]

x = my_generator_function(current_node_test, results_memory_test)
for val in x: 
    print(val)
```
