### EXERCISE EXAM PREPARATION

```python
## Simple matrix - vector multiplication using numpy array

import numpy as np

def Mult(mat, vec):
    
    n_rows = len(vec)
    n_cols = len(mat[0])
    # if the number of row of the vector is not the same as cols in the matrix
    if len(vec) != n_cols:
        raise ValueError("There is an error in the dimension! Check it.")
    

    # Define res
    res = [0.0] * len(mat) 
    for i in range(0, len(mat)):
        for j in range(0, n_rows):
            res[i] += mat[i][j] * vec[j] 
    return res

vec1 = np.array([1,2,3])
mat1 = np.array([[1,0,0],[0,1,1],[0,1,1]])

test = len(mat1[0])
print(test)

Mult(mat1, vec1)
```

```python
def Mult_MV(mat, vec):
    # Define the length of rows and cols for matrix and length of vector
    m_rows = len(mat)
    m_cols = len(mat[0])
    v_len = len(vec)

    # Check whether the dimension is not the same
    if v_len != m_cols:
        raise ValueError("The dimension is not compatible!")

    # Define res
    res = [0.0] * m_rows

    # Do the loop
    for i in range(0, m_rows):
        for j in range(0, m_cols):
            res[i] += mat[i][j] * vec[j]
    return res

Mult_MV(mat1, vec1)
```

### PARENT CLASS
This is just the repetition of the parent class from the previous exercise"

```python
class Mat:
    def __init__(self):
        # initialize number of rows and cols in the matrix
        self.__n_rows = 0
        self.__n_cols = 0

    def reinit(self, n, m):
        """
            Insert the number of rows and columns of the matrix.
        """
        # Check the data type of the number of rows and cols
        if not(isinstance(n, int) and isinstance(m, int)):
            raise TypeError("Incorrect value! You should enter only integer")

        # Make sure that the value is above zero
        if n<=0 or m<=0:
            raise ValueError("Value should greater than 0!")
            
        self.__n_rows = n
        self.__n_cols = m

    def n_rows(self):
        return self.__n_rows

    def n_cols(self):
        return self.__n_cols

    def set(self, i, j, val):
        pass

    def has(self, i, j):
        pass

    def mult(self, vec):
        # Check whether the matrix have the compatible dimension with the vector or not
        if len(vec) != self.__n_cols:
            raise ValueError("The value is not Compatible!")
        # Initiate the results container
        res = [0.0] * self.__n_rows
        # Do the looping
        for i in range(0, self.__n_rows):
            for j in range(0, self.__n_cols):
                res[i] += self(i,j) * vec[j]
        
```

### CHILD CLASS

```python
import random

class DenseMat(Mat):
    def __init__(self):
        super().__init__()
        # This will contain the matrix
        self.__data = []

    def reinit(self, n, m):
        super().reinit(n, m)
        # initiate a random matrix and initiate it as a random data
        self.__data = [random.random() for _ in range(0, n*m)]

    def set(self, i, j, val):
        self.__data[i * super().n_cols() + j] = val

    def __call__(self, i, j):
        return self.__data[i * super().n_cols() + j]

    def write_matrix(self):
        for i in range(0, super().n_cols()):
            res = [0.0] * super().n_rows()
            for j in range(0, super().n_rows()):
                idx = i * super().n_cols() + j
                res[j] = self.__data[idx]
            print(res)
```

### TESTING THE CLASS

```python
A = DenseMat()
A.reinit(2, 2)
A.set(0, 0, 0)
A.set(0, 1, 1)
A.set(1, 0, 0)
A.set(1, 1, 0)
A.write_matrix()
```
